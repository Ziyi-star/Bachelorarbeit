## 1S Segmentation of 100hz

```python
# List of files to segment
files = [
    "../../data/RoadRoughness/Raw/CompactGravel/P1/Accelerometer_filtered_e1_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P2/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/CompactGravel/P2/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P3/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/CompactGravel/P3/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P4/Accelerometer_filtered_e1_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P5/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/CompactGravel/P5/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P6/Accelerometer_filtered_e1_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P7/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/CompactGravel/P7/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P8/Accelerometer_filtered_e1_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P9/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/CompactGravel/P9/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/CompactGravel/P10/Accelerometer_filtered_e1_100hz.csv",
]
```

```python
# Define segmentation parameters
window_size = 100  # 1 second at 100Hz
overlap = 50       # 50% overlap
channels = ['Acc-X', 'Acc-Y', 'Acc-Z']

# Process each file
for file_path in files:
    df = pd.read_csv(file_path)
    
    # Ensure NTP is in datetime format
    df['NTP'] = pd.to_datetime(df['NTP'])
    
    # Apply segmentation
    segments = segment_acceleration_data_overlapping_numpy(
        df,
        window_size=window_size,
        overlap=overlap,
        channels=channels
    )
    
    # Create output filename
    output_path = file_path.replace('.csv', f'_segments_1s_{overlap}overlap.npz')
    # Save segments to NPZ file
    np.savez(output_path, segments=segments)
```
