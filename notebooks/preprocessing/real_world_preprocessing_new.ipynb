```python
def label_curb_scenes(df_data, df_curb, milliseconds=1000, save_path=None):
    """
    Labels curb scenes in the data and extends labels to surrounding data points for validation test.
    
    Args:
        df_data (pd.DataFrame): Original dataframe containing NTP and Acc-Z data
        df_curb (pd.DataFrame): Dataframe containing curb timestamps
        milliseconds (int): Time threshold in milliseconds
        save_path (str, optional): Path to save the CSV file. If None, file won't be saved
        
    Returns:
        pd.DataFrame: DataFrame with labeled curb scenes
    """
    # Create copy with selected columns
    df_selected = df_data[['NTP', 'Acc-X', 'Acc-Y', 'Acc-Z']].copy()
    df_selected['label'] = 0
    
    # Convert NTP columns to datetime
    df_selected['NTP'] = pd.to_datetime(df_selected['NTP'])
    df_curb['Timestamp'] = pd.to_datetime(df_curb['Timestamp'])
    
    # Only get timestamps where Label is not 0, 8, or 9
    curb_timestamps = df_curb[~df_curb['Label'].isin([0, 8, 9])]['Timestamp'].tolist()
    print(f"Number of curb scenes to label: {len(curb_timestamps)}")

    # Use a time threshold approach
    time_threshold = pd.Timedelta(milliseconds=milliseconds)
    for curb_time in curb_timestamps:
        mask = abs(df_selected['NTP'] - curb_time) <= time_threshold
        df_selected.loc[mask, 'label'] = 1
    
    # Save to CSV if path is provided
    if save_path:
        df_selected.to_csv(save_path, index=False)

    return df_selected
```
