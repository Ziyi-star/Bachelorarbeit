## 1s Segmenation of 100hz

```python
# List of files to segment
files = [
    "../../data/RoadRoughness/Raw/Asphalt/P1/Accelerometer_filtered_e1_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P2/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/Asphalt/P2/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P3/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/Asphalt/P3/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P4/Accelerometer_filtered_e1_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P5/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/Asphalt/P5/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P6/Accelerometer_filtered_e1_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P7/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/Asphalt/P7/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P8/Accelerometer_filtered_e1_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P9/Accelerometer_filtered_e1_100hz.csv",
    "../../data/RoadRoughness/Raw/Asphalt/P9/Accelerometer_filtered_e2_100hz.csv",

    "../../data/RoadRoughness/Raw/Asphalt/P10/Accelerometer_filtered_e1_100hz.csv",
]
```

```python
# Define segmentation parameters
window_size = 100  # 1 second at 100Hz
overlap = 50       # 50% overlap
channels = ['Acc-X', 'Acc-Y', 'Acc-Z']

# Process each file
for file_path in files:
    df = pd.read_csv(file_path)
    
    # Ensure NTP is in datetime format
    df['NTP'] = pd.to_datetime(df['NTP'])
    
    # Apply segmentation
    segments = segment_acceleration_data_overlapping_numpy(
        df,
        window_size=window_size,
        overlap=overlap,
        channels=channels
    )
    
    # Create output filename
    output_path = file_path.replace('.csv', f'_segments_1s_{overlap}overlap.npz')
    # Save segments to NPZ file
    np.savez(output_path, segments=segments)
```
